```python
 #3 Which factors are associated with order returns and cancellations?
df["Return_or_Cancelled"] = np.where(
    (df["Return_Status"] == "Yes") |
    (df["Order_Status"].isin(["Returned", "Cancelled"])),
    "Yes",
    "No"
)

df["Risk_Flag"] = np.where(
    df["Return_or_Cancelled"] == "Yes",
    1,
    0
)
```

```python
df["Return_or_Cancelled"].value_counts()

print(
    "Overall return/cancellation rate:",
    round(df["Risk_Flag"].mean() * 100, 2),
    "%"
)
```

```python
def risk_summary(column_name):
    summary = df.groupby(column_name)["Risk_Flag"].agg(
        ["count", "sum", "mean"]
    )

    summary = summary.rename(columns={
        "count": "Total_Orders",
        "sum": "Return_or_Cancelled_Orders",
        "mean": "Return_or_Cancelled_Rate"
    })

    summary["Return_or_Cancelled_Rate"] = (
        summary["Return_or_Cancelled_Rate"] * 100
    ).round(2)

    return summary.sort_values(
        "Return_or_Cancelled_Rate",
        ascending=False
    )
```

```python
payment_risk = risk_summary("Payment_Method")
payment_risk
```

```python
device_risk = risk_summary("Device")
device_risk
```

```python
category_risk = risk_summary("Category")
category_risk
```

```python
region_risk = risk_summary("Region")
region_risk
```

```python
factors = [
    "Payment_Method",
    "Device",
    "Category",
    "Region"
]

fig, axes = plt.subplots(2, 2, figsize=(15, 10))

for factor, axis in zip(factors, axes.flat):
    
    risk_rate = df.groupby(factor)["Risk_Flag"].mean() * 100
    risk_rate = risk_rate.sort_values(ascending=False)
    
    risk_rate.plot(
        kind="bar",
        ax=axis,
        color="tomato"
    )
    
    axis.set_title(
        f"Return/Cancellation Rate by {factor}"
    )
    axis.set_xlabel("")
    axis.set_ylabel("Rate (%)")
    axis.tick_params(axis="x", rotation=35)

plt.tight_layout()
plt.show()
```

Returns and cancellations

Overall return/cancellation rate: 24.8%.

Among known payment methods, UPI has the highest rate: 26.45%

Electronics has the highest category rate: 26.34%

Central region has the highest regional rate: 25.49%

Device rates are almost identical, so device does not appear to be an important factor.

Do not treat "Unknown" payment method as a real payment-method insight.

Focus return-reduction investigation on UPI transactions, Electronics, and Central region; the differences are present but moderate.

```python
#Problem 4

#Does customer segment influence order value and profitability?
segment_summary = df.groupby("Customer_Segment").agg(
    Total_Orders=("Order_ID", "count"),
    Average_Net_Sales=("Net_Sales", "mean"),
    Average_Profit=("Profit", "mean"),
    Average_Profit_Margin=("Profit_Margin_Pct", "mean")
).round(2)

segment_summary
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.barplot(
    data=segment_summary.reset_index(),
    x="Customer_Segment",
    y="Average_Net_Sales",
    ax=axes[0],
    color="steelblue"
)

axes[0].set_title("Average Net Sales by Customer Segment")
axes[0].set_xlabel("Customer Segment")
axes[0].set_ylabel("Average Net Sales")

sns.barplot(
    data=segment_summary.reset_index(),
    x="Customer_Segment",
    y="Average_Profit",
    ax=axes[1],
    color="seagreen"
)

axes[1].set_title("Average Profit by Customer Segment")
axes[1].set_xlabel("Customer Segment")
axes[1].set_ylabel("Average Profit")

plt.tight_layout()
plt.show()
```
